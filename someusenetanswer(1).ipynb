```python
group_pattern = re.compile(r"^(alt|comp|soc|rec|talk|news|sci|misc|humanities|uk|aus|de|fr|can|nl)\.[\w\.\-]+")
match_sender = re.compile(r"^From:")
match_subject = re.compile(r"^Subject:")
match_ref = re.compile(r"^References:")
match_msgid = re.compile(r"^Message-ID:")
match_xref = re.compile(r"^Xref:")

cleaned_df = []
columns = ['index', 'sender', 'subject', 'references', 'groups', 'body', 'msg_id']
msg_parts = df_messages_new["msg_parts"].astype(str)
i = 0

while i < len(msg_parts):
    row = {'index': i, 'sender': None, 'subject': None, 'references': None, 'groups': None, 'body': "", 'msg_id': None}
    while i < len(msg_parts):
        line = msg_parts.iloc[i]

        if match_sender.match(line):
            row['sender'] = line.replace("From:", "").strip()

        elif match_subject.match(line):
            row['subject'] = line.replace("Subject:", "").strip()

        elif match_ref.match(line):
            row['references'] = line.replace("References:", "").strip()

        elif match_msgid.match(line):
            row['msg_id'] = line.replace("Message-ID:", "").strip()

        elif match_xref.match(line):
            groups = group_pattern.findall(line)
            row['groups'] = ",".join(groups) if groups else None

        elif i+1 < len(msg_parts) and match_sender.match(msg_parts.iloc[i+1]):
            break

        else:
            row['body'] += line + "\n"

        i += 1

    cleaned_df.append(row)
    i += 1  


df_cleaned = pd.DataFrame(cleaned_df, columns=columns)
```

```python
# Step 1: Build a lookup dictionary for msg_id → row index
msgid_to_index = df_cleaned.groupby('msg_id')['index'].apply(list).to_dict()

# Step 2: Build a reference mapping tree (msg_id → list of replies)
from collections import defaultdict

thread_tree = defaultdict(list)

for idx, row in df_cleaned.iterrows():
    ref = row['references']
    if ref and ref in msgid_to_index:
        thread_tree[ref].append(row['msg_id'])  # this message is a reply to `ref`
```

```python
# 1. Build child-to-parent map
msg_to_parent = {}

for parent, replies in thread_tree.items():
    for child in replies:
        msg_to_parent[child] = parent

# 2. Calculate depth recursively
depths = {}

def get_depth(msg_id, visited=None):
    if visited is None:
        visited = set()
    if msg_id in visited:
        return 0  # break circular loop

    visited.add(msg_id)

    if msg_id not in msg_to_parent:
        return 0  # root node

    parent = msg_to_parent[msg_id]

    if parent not in depths:
        depths[parent] = get_depth(parent, visited)

    return depths[parent] + 1

# 3. Apply depth calculation to all msg_ids
for msg_id in df_cleaned['msg_id']:
    depths[msg_id] = get_depth(msg_id)

# 4. Add depth info to DataFrame
df_cleaned['depth'] = df_cleaned['msg_id'].map(depths)
```
